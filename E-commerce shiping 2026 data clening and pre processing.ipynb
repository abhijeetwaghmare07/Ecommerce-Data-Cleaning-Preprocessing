# Outlier Detection

```python
numerical_columns = [
    "product_weight_kg",
    "order_value_usd",
    "distance_km",
    "promised_delivery_days",
    "actual_delivery_days",
    "delivery_delay_days",
    "shipping_cost_usd",
    "customer_rating",
    "delivery_attempts",
    "warehouse_processing_hours"
]

display(df_copy[numerical_columns].describe())
```

```python
# Define numerical columns for outlier detection
numerical_columns = [
    "product_weight_kg",
    "order_value_usd",
    "distance_km",
    "promised_delivery_days",
    "actual_delivery_days",
    "delivery_delay_days",
    "shipping_cost_usd",
    "customer_rating",
    "delivery_attempts",
    "warehouse_processing_hours"
]

# Calculate outliers using IQR
outlier_summary = []

for column in numerical_columns:
    Q1 = df_copy[column].quantile(0.25)
    Q3 = df_copy[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_copy[
        (df_copy[column] < lower_limit) |
        (df_copy[column] > upper_limit)
    ]

    outlier_summary.append({
        "Column": column,
        "Lower Limit": round(lower_limit, 2),
        "Upper Limit": round(upper_limit, 2),
        "Outlier Count": len(outliers)
    })

outlier_df_copy = pd.DataFrame(outlier_summary)

display(outlier_df_copy)
```

# Outlier Treatment
Purpose:
Instead of deleting valid logistics records, extreme values will be capped at the calculated IQR lower and upper limits. This reduces the influence of extreme observations while preserving the original records.

```python
outlier_columns = [
    "product_weight_kg",
    "order_value_usd",
    "distance_km",
    "actual_delivery_days",
    "delivery_delay_days",
    "shipping_cost_usd",
    "warehouse_processing_hours"
]

for column in outlier_columns:
    Q1 = df_copy[column].quantile(0.25)
    Q3 = df_copy[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df_copy[column] = df_copy[column].clip(
        lower=lower_limit,
        upper=upper_limit
    )

print("Outlier treatment completed successfully.")
```

# Normalization/Standardization add

## 8. Data Normalization / Standardization

Normalization is performed to bring numerical features to a comparable scale.
StandardScaler is used to standardize numerical variables by removing the mean
and scaling them to unit variance.

This is useful because logistics features such as distance, shipping cost,
delivery days, and order value have different numerical ranges.

```python
from sklearn.preprocessing import StandardScaler

# Numerical columns selected for standardization
scaling_columns = [
    "distance_km",
    "shipping_cost_usd",
    "actual_delivery_days",
    "delivery_delay_days"
]

# Create scaler
scaler = StandardScaler()

# Apply standardization
df_copy[scaling_columns] = scaler.fit_transform(
    df_copy[scaling_columns]
)

print("Standardization completed successfully.")
```
